# Bronze Data Ingestion

This notebook handles the ingestion of raw e-commerce data into the Bronze layer of the Unity Catalog. Data is preserved as-is, without transformations, to serve as the foundation for downstream Silver and Gold layers.

## Objective

The Bronze layer stores raw data exactly as received from the source — no cleaning, no transformation. This ensures traceability and reproducibility: if a downstream transformation fails or business rules change, the original data remains intact. This notebook sets up the Bronze schema in Unity Catalog and validates the ingested tables in the `projetoportfolio.bronze` schema.

## Data Sources

| Source File | Bronze Table | Purpose |
|---|---|---|
| `vendas.parquet` | `projetoportfolio.bronze.vendas` | Sales transactions |
| `produtos.parquet` | `projetoportfolio.bronze.produtos` | Product catalog |
| `clientes.parquet` | `projetoportfolio.bronze.clientes` | Customer records |
| `preco_competidores.parquet` | `projetoportfolio.bronze.preco_competidores` | Competitor pricing |

Source Parquet files were loaded into Delta tables using Spark, preserving the raw data without transformation.

## Data Ingestion

### Unity Catalog Setup

Create the `projetoportfolio` catalog and `bronze` schema in Unity Catalog.

```python
spark.sql("CREATE CATALOG IF NOT EXISTS projetoportfolio")
spark.sql("CREATE SCHEMA IF NOT EXISTS projetoportfolio.bronze")
```

### Bronze Table Inspection

Display sample records from each Bronze table to verify the ingested data.

```python
for tabela in ["vendas", "produtos", "clientes", "preco_competidores"]:
    print(f"\n=== projetoportfolio.bronze.{tabela} ===")
    display(spark.table(f"projetoportfolio.bronze.{tabela}").limit(5))
```

## Bronze Tables

| Table | Source | Purpose |
|---|---|---|

## Data Validation

Verify row counts across all Bronze tables.

```python
%sql
SELECT 'vendas' AS tabela, COUNT(*) AS linhas FROM projetoportfolio.bronze.vendas
UNION ALL SELECT 'produtos', COUNT(*) FROM projetoportfolio.bronze.produtos
UNION ALL SELECT 'clientes', COUNT(*) FROM projetoportfolio.bronze.clientes
UNION ALL SELECT 'preco_competidores', COUNT(*) FROM projetoportfolio.bronze.preco_competidores
```
